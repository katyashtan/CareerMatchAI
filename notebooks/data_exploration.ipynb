```python
import pandas as pd


df = pd.read_csv('../data/raw/ai_jobs_global.csv')

missing_values = df.isnull().sum()

print(f"Size of dataset: {df.shape}")
print('Number of missing values in each column:')
print(missing_values)
print('Type  of each column:')
print(df.dtypes)
```

```python
df_clean = df.drop(columns=['country', 'currency', 'source'])
df_clean[['company', 'posted_date']] = df_clean[['company', 'posted_date']].fillna("Unknow")

df_clean["desc_len"] = df_clean['job_description'].str.len()
print(df_clean["desc_len"].describe())
df_clean = df_clean[df_clean['desc_len'] > 100]
df_clean = df_clean.drop(columns=['desc_len'])

missing_values = df_clean.isnull().sum()

print(f"Size of dataset: {df_clean.shape}")
print(f"Sum of dublicates: {df_clean.duplicated().sum()}")
print('Number of missing values in each column:')
print(missing_values)

df_clean.to_csv('../data/processed/clean_data.csv', index=False)
```
